```python
def process_xy_positions_efficient(xy_positions, image_path, quad, 
                                 output_csv="flux_results.csv"):
    """
    Efficiently process multiple x,y positions by loading the image once
    
    Parameters:
    -----------
    xy_positions : array-like
        Array of (x, y) coordinates, shape (N, 2)
    image_path : str
        Path to FITS file
    quad : int
        Quadrant number
    tracking : bool
        Whether to track streak position
    output_csv : str
        Output CSV filename
        
    Returns:
    --------
    nothing, saves light curves
    """
    xy_positions = np.array(xy_positions)
    
    filename = image_path[-47:-8]
    
    print(f"Loading FITS file: {image_path}")
    print(f"Processing {len(xy_positions)} positions...")
    
    # Load image once
    try:
        hdul = fits.open(image_path)
        file1_array = hdul[quad].data
        hdul.close()
        
        # Calculate background once
        print("Calculating background...")
        background = rff.get_image_background(file1_array)
        imagenobackground = file1_array - background
        
        print(f"Image shape: {imagenobackground.shape}")
        print(f"Background value: {background:.2f}")
        
    except Exception as e:
        print(f"Error loading FITS file: {e}")
        return None
    
    # Process all positions
    all_results = []
    
    for i, (x_pos, y_pos) in enumerate(xy_positions):
        try:
            tot_2,tot_3,tot_4,tot_5, maxcols = get_tot_from_xy_efficient(
                imagenobackground, int(x_pos), int(y_pos), tracking=True)
            
            # Create DataFrame for this position
            position_df = pd.DataFrame({
                'position_id': i,
                'x_input': x_pos,
                'y_input': y_pos,
                'row_index': range(len(tot)),
                'flux_hap_2': tot_2,
                'flux_hap_3': tot_3,
                'flux_hap_4': tot_4,
                'flux_hap_5': tot_5,
                'max_col': maxcols
            })
            
            all_results.append(position_df)
            

            individual_filename = f"/lustre/lrspec/ZTF_data/lightcurves/1020/lightcurve_{filename}_{x_pos}_{y_pos}.csv"
            position_df.to_csv(individual_filename, index=False)

                
        except Exception as e:
            print(f"Error processing position {i} ({x_pos}, {y_pos}): {e}")
            continue
    
    return
```

```python
def get_tot_from_xy_efficient(imagenobackground, x_pos, y_pos, tracking=True):
    """
    Extract flux from a single x,y position using pre-loaded image data
    
    Parameters:
    -----------
    imagenobackground : array
        Pre-loaded and background-subtracted image data
    x_pos : int
        X coordinate (column)
    y_pos : int
        Y coordinate (row)
    half_aperture : int
        Half-width of aperture
    tracking : bool
        Whether to track streak position
        
    Returns:
    --------
    tot : array
        Flux values for each row
    maxcols : array
        Column positions used for each row
    """
    if tracking == False:
        # Fixed position approach
        zeropoint = x_pos - 7
        search_start = max(0, zeropoint)
        search_end = min(imagenobackground.shape[1], zeropoint + 14)
        
        if search_end > search_start and y_pos - 10 >= 0 and y_pos + 10 < imagenobackground.shape[0]:
            averages = np.average(
                imagenobackground[y_pos - 10 : y_pos + 10, search_start : search_end], axis=0
            )
            maxind = np.argmax(averages)
            maxcol = search_start + maxind
        else:
            maxcol = x_pos
            
        tot = np.zeros(imagenobackground.shape[0])
        
        for r in range(imagenobackground.shape[0]):
            combined = imagenobackground[r, maxcol] if maxcol < imagenobackground.shape[1] else 0
            for i in range(1, half_aperture + 1):
                if maxcol - i >= 0:
                    combined += imagenobackground[r, maxcol - i]
                if maxcol + i < imagenobackground.shape[1]:
                    combined += imagenobackground[r, maxcol + i]
            tot[r] = combined
        
        return tot, np.full(len(tot), maxcol)
    
    else:
        # Tracking approach
        tot_2 = np.zeros(imagenobackground.shape[0])
        tot_3 = np.zeros(imagenobackground.shape[0])
        tot_4 = np.zeros(imagenobackground.shape[0])
        tot_5 = np.zeros(imagenobackground.shape[0])
        maxcols = np.zeros(imagenobackground.shape[0])
        zeropoint = x_pos - 10
        
        for r in range(imagenobackground.shape[0]):
            search_start = max(0, zeropoint)
            search_end = min(imagenobackground.shape[1], zeropoint + 20)
            
            if search_end > search_start:
                maxcol = np.argmax(imagenobackground[r, search_start:search_end]) + search_start
            else:
                maxcol = x_pos
                
            maxcols[r] = maxcol
            combined = imagenobackground[r, maxcol] if maxcol < imagenobackground.shape[1] else 0
            
            half_aperture = 2
            for i in range(1, half_aperture + 1):
                if maxcol - i >= 0:
                    combined += imagenobackground[r, maxcol - i]
                if maxcol + i < imagenobackground.shape[1]:
                    combined += imagenobackground[r, maxcol + i]
            
            tot_2[r] = combined
            
            half_aperture = 3
            for i in range(1, half_aperture + 1):
                if maxcol - i >= 0:
                    combined += imagenobackground[r, maxcol - i]
                if maxcol + i < imagenobackground.shape[1]:
                    combined += imagenobackground[r, maxcol + i]
            
            tot_3[r] = combined
            
            half_aperture = 4
            for i in range(1, half_aperture + 1):
                if maxcol - i >= 0:
                    combined += imagenobackground[r, maxcol - i]
                if maxcol + i < imagenobackground.shape[1]:
                    combined += imagenobackground[r, maxcol + i]
            
            tot_4[r] = combined
            
            half_aperture = 5
            for i in range(1, half_aperture + 1):
                if maxcol - i >= 0:
                    combined += imagenobackground[r, maxcol - i]
                if maxcol + i < imagenobackground.shape[1]:
                    combined += imagenobackground[r, maxcol + i]
            
            tot_5[r] = combined
            
            
        return tot_2,tot_3,tot_4,tot_5, maxcols
    
def print_summary_stats(results_df, xy_positions):
    """Print summary statistics"""
    print(f"\nSummary Statistics:")
    print(f"Total positions processed: {len(xy_positions)}")
    print(f"Total flux measurements: {len(results_df)}")
    print(f"Average flux per position: {results_df.groupby('position_id')['flux'].mean().mean():.2f}")
    print(f"Max flux: {results_df['flux'].max():.2f}")
    print(f"Min flux: {results_df['flux'].min():.2f}")

def create_summary_csv(results_df, output_csv="flux_summary.csv"):
    """
    Create a summary CSV with key statistics for each position
    """
    if results_df is None:
        return
    
    summary = results_df.groupby(['position_id', 'x_input', 'y_input']).agg({
        'flux': ['mean', 'max', 'min', 'std', 'sum'],
        'max_col': ['mean', 'std']
    }).round(3)
    
    # Flatten column names
    summary.columns = ['_'.join(col).strip() for col in summary.columns.values]
    summary = summary.reset_index()
    
    summary.to_csv(output_csv, index=False)
    print(f"Summary statistics saved to: {output_csv}")
    
    return summary
```

```python




def process_xy_positions_efficient(xy_positions, image_path, quad, half_aperture=2, tracking=True, 
                                 output_csv="flux_results.csv", save_individual_files=False, 
                                 batch_size=None):
    """
    Efficiently process multiple x,y positions by loading the image once
    
    Parameters:
    -----------
    xy_positions : array-like
        Array of (x, y) coordinates, shape (N, 2)
    image_path : str
        Path to FITS file
    quad : int
        Quadrant number
    half_aperture : int
        Half-width of aperture
    tracking : bool
        Whether to track streak position
    output_csv : str
        Output CSV filename
    save_individual_files : bool
        Whether to save individual CSV files for each position
    batch_size : int or None
        If specified, save results in batches to manage memory
        
    Returns:
    --------
    results_df : DataFrame
        Combined results DataFrame
    """
    xy_positions = np.array(xy_positions)
    
    print(f"Loading FITS file: {image_path}")
    print(f"Processing {len(xy_positions)} positions...")
    
    # Load image once
    try:
        hdul = fits.open(image_path)
        file1_array = hdul[quad].data
        hdul.close()
        
        # Calculate background once
        print("Calculating background...")
        background = get_image_background(file1_array)
        imagenobackground = file1_array - background
        
        print(f"Image shape: {imagenobackground.shape}")
        print(f"Background value: {background:.2f}")
        
    except Exception as e:
        print(f"Error loading FITS file: {e}")
        return None
    
    # Process all positions
    all_results = []
    batch_count = 0
    
    for i, (x_pos, y_pos) in enumerate(xy_positions):
        try:
            if i % 100 == 0:  # Progress update every 100 positions
                print(f"Processing position {i+1}/{len(xy_positions)}: ({x_pos}, {y_pos})")
            
            tot, maxcols = get_tot_from_xy_efficient(
                imagenobackground, int(x_pos), int(y_pos), half_aperture, tracking
            )
            
            # Create DataFrame for this position
            position_df = pd.DataFrame({
                'position_id': i,
                'x_input': x_pos,
                'y_input': y_pos,
                'row_index': range(len(tot)),
                'flux': tot,
                'max_col': maxcols
            })
            
            all_results.append(position_df)
            
            # Save individual file if requested
            if save_individual_files:
                individual_filename = f"position_{i}_{x_pos}_{y_pos}.csv"
                position_df.to_csv(individual_filename, index=False)
            
            # Handle batching if specified
            if batch_size is not None and len(all_results) >= batch_size:
                batch_df = pd.concat(all_results, ignore_index=True)
                batch_filename = f"batch_{batch_count}_{output_csv}"
                batch_df.to_csv(batch_filename, index=False)
                print(f"Saved batch {batch_count} to: {batch_filename}")
                all_results = []
                batch_count += 1
                
        except Exception as e:
            print(f"Error processing position {i} ({x_pos}, {y_pos}): {e}")
            continue
    
    # Combine all results
    if all_results:
        results_df = pd.concat(all_results, ignore_index=True)
        
        # Save final results
        results_df.to_csv(output_csv, index=False)
        print(f"Saved combined results to: {output_csv}")
        
        # Print summary statistics
        print_summary_stats(results_df, xy_positions)
        
        return results_df
    else:
        print("No results to save!")
        return None

def process_xy_positions_vectorized(xy_positions, image_path, quad, half_aperture=2, 
                                   output_csv="flux_results.csv", tracking_window=20):
    """
    Highly optimized vectorized version for processing many positions
    Note: This version uses a simplified tracking approach for speed
    
    Parameters:
    -----------
    xy_positions : array-like
        Array of (x, y) coordinates, shape (N, 2)
    image_path : str
        Path to FITS file
    quad : int
        Quadrant number
    half_aperture : int
        Half-width of aperture
    output_csv : str
        Output CSV filename
    tracking_window : int
        Window size for finding peak in each row
        
    Returns:
    --------
    results_df : DataFrame
        Combined results DataFrame
    """
    xy_positions = np.array(xy_positions)
    
    print(f"Loading FITS file: {image_path}")
    print(f"Vectorized processing of {len(xy_positions)} positions...")
    
    # Load image once
    try:
        hdul = fits.open(image_path)
        file1_array = hdul[quad].data
        hdul.close()
        
        background = get_image_background(file1_array)
        imagenobackground = file1_array - background
        
    except Exception as e:
        print(f"Error loading FITS file: {e}")
        return None
    
    # Prepare output arrays
    n_rows = imagenobackground.shape[0]
    n_positions = len(xy_positions)
    
    # Process all positions in vectorized manner
    all_results = []
    
    for i, (x_pos, y_pos) in enumerate(xy_positions):
        if i % 100 == 0:
            print(f"Processing position {i+1}/{n_positions}")
        
        try:
            x_pos, y_pos = int(x_pos), int(y_pos)
            
            # Create aperture indices
            aperture_cols = np.arange(x_pos - half_aperture, x_pos + half_aperture + 1)
            aperture_cols = aperture_cols[(aperture_cols >= 0) & (aperture_cols < imagenobackground.shape[1])]
            
            if len(aperture_cols) == 0:
                continue
            
            # For each row, find the peak within tracking window and sum aperture
            tot = np.zeros(n_rows)
            maxcols = np.zeros(n_rows)
            
            for r in range(n_rows):
                # Define search window
                search_start = max(0, x_pos - tracking_window // 2)
                search_end = min(imagenobackground.shape[1], x_pos + tracking_window // 2)
                
                if search_end > search_start:
                    # Find peak in search window
                    peak_idx = np.argmax(imagenobackground[r, search_start:search_end])
                    peak_col = search_start + peak_idx
                else:
                    peak_col = x_pos
                
                maxcols[r] = peak_col
                
                # Sum aperture around peak
                aperture_around_peak = np.arange(peak_col - half_aperture, peak_col + half_aperture + 1)
                aperture_around_peak = aperture_around_peak[
                    (aperture_around_peak >= 0) & (aperture_around_peak < imagenobackground.shape[1])
                ]
                
                if len(aperture_around_peak) > 0:
                    tot[r] = np.sum(imagenobackground[r, aperture_around_peak])
                else:
                    tot[r] = 0
            
            # Create DataFrame for this position
            position_df = pd.DataFrame({
                'position_id': i,
                'x_input': x_pos,
                'y_input': y_pos,
                'row_index': range(len(tot)),
                'flux': tot,
                'max_col': maxcols
            })
            
            all_results.append(position_df)
            
        except Exception as e:
            print(f"Error processing position {i} ({x_pos}, {y_pos}): {e}")
            continue
    
    # Combine and save results
    if all_results:
        results_df = pd.concat(all_results, ignore_index=True)
        results_df.to_csv(output_csv, index=False)
        print(f"Saved vectorized results to: {output_csv}")
        
        print_summary_stats(results_df, xy_positions)
        return results_df
    else:
        print("No results to save!")
        return None





# Example usage functions
def example_usage_efficient():
    """
    Example using the efficient version
    """
    # Define positions
    xy_positions = [
        (100, 500), (150, 520), (200, 540), (250, 560), (300, 580),
        (350, 600), (400, 620), (450, 640), (500, 660), (550, 680)
    ]
    
    # Process efficiently (loads image once)
    results_df = process_xy_positions_efficient(
        xy_positions=xy_positions,
        image_path="your_image_file.fits.fz",
        quad=1,
        half_aperture=2,
        tracking=True,
        output_csv="efficient_flux_results.csv"
    )
    
    # Create summary
    if results_df is not None:
        summary_df = create_summary_csv(results_df, "efficient_flux_summary.csv")

def example_usage_vectorized():
    """
    Example using the vectorized version (fastest for many positions)
    """
    # For many positions, use vectorized version
    xy_positions = np.random.randint(50, 500, size=(1000, 2))  # 1000 random positions
    
    results_df = process_xy_positions_vectorized(
        xy_positions=xy_positions,
        image_path="your_image_file.fits.fz",
        quad=1,
        half_aperture=2,
        output_csv="vectorized_flux_results.csv"
    )

def load_xy_positions_from_fits(bintable_path):
    """
    Load X,Y positions from FITS binary table
    
    Parameters:
    -----------
    bintable_path : str
        Path to FITS file containing binary table with X,Y coordinates
        
    Returns:
    --------
    xy_positions : array
        Array of (x, y) coordinates
    """
    try:
        with fits.open(bintable_path) as hdul:
            # Try to find the binary table (usually extension 1, but could vary)
            table_data = None
            for i, hdu in enumerate(hdul):
                if hasattr(hdu, 'data') and hdu.data is not None:
                    if hasattr(hdu.data, 'dtype') and len(hdu.data.dtype.names or []) > 0:
                        table_data = hdu.data
                        print(f"Found binary table in extension {i}")
                        break
            
            if table_data is None:
                print(f"No binary table found in {bintable_path}")
                return None
            
            # Extract X,Y coordinates
            # Handle different possible column names
            x_col = None
            y_col = None
            
            for col_name in table_data.dtype.names:
                if col_name.upper() in ['X', 'X_POS', 'XPOS', 'COL']:
                    x_col = col_name
                elif col_name.upper() in ['Y', 'Y_POS', 'YPOS', 'ROW']:
                    y_col = col_name
            
            if x_col is None or y_col is None:
                print(f"Could not find X,Y columns in {bintable_path}")
                print(f"Available columns: {table_data.dtype.names}")
                return None
            
            # Extract positions
            x_vals = table_data[x_col]
            y_vals = table_data[y_col]
            xy_positions = np.column_stack([x_vals, y_vals])
            
            print(f"Loaded {len(xy_positions)} positions from {bintable_path}")
            print(f"X range: {x_vals.min():.1f} - {x_vals.max():.1f}")
            print(f"Y range: {y_vals.min():.1f} - {y_vals.max():.1f}")
            
            return xy_positions
            
    except Exception as e:
        print(f"Error loading positions from {bintable_path}: {e}")
        return None

def process_image_with_corresponding_positions(image_path, positions_path, quad, 
                                             half_aperture=2, output_csv=None):
    """
    Process a single image with its corresponding positions file
    
    Parameters:
    -----------
    image_path : str
        Path to the image FITS file
    positions_path : str
        Path to the positions FITS binary table
    quad : int
        Quadrant to process
    half_aperture : int
        Half-width of aperture
    output_csv : str or None
        Output CSV filename (auto-generated if None)
        
    Returns:
    --------
    results_df : DataFrame
        Results dataframe
    """
    # Load positions
    xy_positions = load_xy_positions_from_fits(positions_path)
    if xy_positions is None:
        return None
    
    # Generate output filename if not provided
    if output_csv is None:
        base_name = os.path.basename(image_path).replace(".fits.fz", "").replace(".fits", "")
        output_csv = f"flux_results_{base_name}.csv"
    
    # Process with vectorized version
    results_df = process_xy_positions_vectorized(
        xy_positions=xy_positions,
        image_path=image_path,
        quad=quad,
        half_aperture=half_aperture,
        output_csv=output_csv
    )
    
    return results_df

def process_multiple_images_with_positions(image_dir, positions_dir, quad=1, 
                                          half_aperture=2, output_dir="./"):
    """
    Process multiple images, each with corresponding position files
    
    Parameters:
    -----------
    image_dir : str
        Directory containing image FITS files
    positions_dir : str
        Directory containing position FITS files
    quad : int
        Quadrant to process (default 1)
    half_aperture : int
        Half-width of aperture
    output_dir : str
        Directory to save output CSV files
        
    Returns:
    --------
    processed_files : list
        List of successfully processed files
    """
    import glob
    
    # Find all image files
    image_files = glob.glob(os.path.join(image_dir, "*.fits.fz"))
    image_files.extend(glob.glob(os.path.join(image_dir, "*.fits")))
    
    if not image_files:
        print(f"No FITS files found in {image_dir}")
        return []
    
    print(f"Found {len(image_files)} image files to process...")
    
    processed_files = []
    
    for i, image_path in enumerate(image_files):
        # Get base filename without extension
        base_name = os.path.basename(image_path)
        base_name = base_name.replace(".fits.fz", "").replace(".fits", "")
        
        # Look for corresponding position file
        position_file = None
        for ext in [".fits", ".fits.fz"]:
            potential_path = os.path.join(positions_dir, base_name + ext)
            if os.path.exists(potential_path):
                position_file = potential_path
                break
        
        if position_file is None:
            print(f"Warning: No position file found for {base_name}")
            continue
        
        print(f"\n--- Processing {i+1}/{len(image_files)}: {base_name} ---")
        print(f"Image: {image_path}")
        print(f"Positions: {position_file}")
        
        # Generate output path
        output_csv = os.path.join(output_dir, f"flux_results_{base_name}.csv")
        
        # Process this image+positions pair
        results_df = process_image_with_corresponding_positions(
            image_path=image_path,
            positions_path=position_file,
            quad=quad,
            half_aperture=half_aperture,
            output_csv=output_csv
        )
        
        if results_df is not None:
            processed_files.append(base_name)
            
            # Also create summary file
            summary_csv = os.path.join(output_dir, f"flux_summary_{base_name}.csv")
            create_summary_csv(results_df, summary_csv)
        else:
            print(f"Failed to process {base_name}")
    
    print(f"\nCompleted! Successfully processed {len(processed_files)} files:")
    for filename in processed_files:
        print(f"  - {filename}")
    
    return processed_files

def example_usage_with_bintables():
    """
    Example usage with FITS binary tables
    """
    # Process a single image with its position file
    results = process_image_with_corresponding_positions(
        image_path="/path/to/image.fits.fz",
        positions_path="/path/to/positions.fits",
        quad=1,
        half_aperture=2
    )
    
    # Or process multiple images with their corresponding position files
    processed = process_multiple_images_with_positions(
        image_dir="/path/to/images/",
        positions_dir="/path/to/positions/",
        quad=1,
        half_aperture=2,
        output_dir="/path/to/output/"
    )

if __name__ == "__main__":
    # Example usage for your workflow
    example_usage_with_bintables()
    
    # Or run the batch processing
    # process_multiple_images_with_positions(
    #     image_dir="/lustre/lrspec/ZTF_data/science_extractions/to_extract/1020_minus_extras/",
    #     positions_dir="/path/to/your/position/files/",
    #     quad=1,
    #     output_dir="/path/to/output/"
    # )
```

```python
def load_xy_positions_from_fits(bintable_path):
    """
    Load X,Y positions from FITS binary table
    
    Parameters:
    -----------
    bintable_path : str
        Path to FITS file containing binary table with X,Y coordinates
        
    Returns:
    --------
    xy_positions : array
        Array of (x, y) coordinates
    """
    try:
        with fits.open(bintable_path) as hdul:
            # Try to find the binary table (usually extension 1, but could vary)
            table_data = None
            for i, hdu in enumerate(hdul):
                if hasattr(hdu, 'data') and hdu.data is not None:
                    if hasattr(hdu.data, 'dtype') and len(hdu.data.dtype.names or []) > 0:
                        table_data = hdu.data
                        print(f"Found binary table in extension {i}")
                        break
            
            if table_data is None:
                print(f"No binary table found in {bintable_path}")
                return None
            
            # Extract X,Y coordinates
            # Handle different possible column names
            x_col = None
            y_col = None
            
            for col_name in table_data.dtype.names:
                if col_name.upper() in ['X', 'X_POS', 'XPOS', 'COL']:
                    x_col = col_name
                elif col_name.upper() in ['Y', 'Y_POS', 'YPOS', 'ROW']:
                    y_col = col_name
            
            if x_col is None or y_col is None:
                print(f"Could not find X,Y columns in {bintable_path}")
                print(f"Available columns: {table_data.dtype.names}")
                return None
            
            # Extract positions
            x_vals = table_data[x_col]
            y_vals = table_data[y_col]
            xy_positions = np.column_stack([x_vals, y_vals])
            
            print(f"Loaded {len(xy_positions)} positions from {bintable_path}")
            print(f"X range: {x_vals.min():.1f} - {x_vals.max():.1f}")
            print(f"Y range: {y_vals.min():.1f} - {y_vals.max():.1f}")
            
            return xy_positions
            
    except Exception as e:
        print(f"Error loading positions from {bintable_path}: {e}")
        return None
```
