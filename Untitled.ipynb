```python
from astropy.coordinates import SkyCoord, EarthLocation, AltAz
from astropy.time import Time
import astropy.units as u
```

```python
palomar = EarthLocation(lat=33.356*u.deg, lon=-116.863*u.deg, height=1712*u.m)
c = SkyCoord(298.26936*u.deg, 33.35*u.deg)
t = Time("2022-10-20 03:19:05") + (930/2)*u.s   # mid-point of a row
z_686 = 90*u.deg - c.transform_to(AltAz(obstime=t, location=palomar)).alt
```
